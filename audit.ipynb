```python
r = requests.get("https://lldev.thespacedevs.com/2.3.0/launches/previous/",
                 params={"limit": 100})
sample = r.json()["results"]

no_mission = 0
no_orbit = 0
orbits = {}

for l in sample:
    m = l.get("mission")
    if not m:
        no_mission += 1
        continue
    o = m.get("orbit")
    if not o:
        no_orbit += 1
        continue
    orbits[o["name"]] = orbits.get(o["name"], 0) + 1

print("no mission record:", no_mission)
print("mission but no orbit:", no_orbit)
for name, n in sorted(orbits.items(), key=lambda x: -x[1]):
    print(f"{n:4}  {name}")
```

```python
import time

all_configs = []
url = "https://lldev.thespacedevs.com/2.3.0/launcher_configurations/"
params = {"limit": 100}

while url:
    r = requests.get(url, params=params)
    data = r.json()
    all_configs.extend(data["results"])
    url = data["next"]
    params = None

print("fetched:", len(all_configs))

leo_null = sum(1 for c in all_configs if c.get("leo_capacity") is None)
leo_zero = sum(1 for c in all_configs if c.get("leo_capacity") == 0)
leo_pos  = sum(1 for c in all_configs if (c.get("leo_capacity") or 0) > 0)
placeholders = sum(1 for c in all_configs if c.get("is_placeholder"))

print("leo_capacity null:", leo_null)
print("leo_capacity zero:", leo_zero)
print("leo_capacity positive:", leo_pos)
print("placeholders:", placeholders)
```

```python
import time

all_configs = []
url = "https://lldev.thespacedevs.com/2.3.0/launcher_configurations/"
params = {"limit": 100, "mode": "detailed"}

while url:
    r = requests.get(url, params=params)
    data = r.json()
    all_configs.extend(data["results"])
    url = data["next"]
    params = None

print("fetched:", len(all_configs))

leo_null = sum(1 for c in all_configs if c.get("leo_capacity") is None)
leo_zero = sum(1 for c in all_configs if c.get("leo_capacity") == 0)
leo_pos  = sum(1 for c in all_configs if (c.get("leo_capacity") or 0) > 0)
apogee_pos = sum(1 for c in all_configs if (c.get("apogee") or 0) > 0)
placeholders = sum(1 for c in all_configs if c.get("is_placeholder"))

print("leo_capacity null:", leo_null)
print("leo_capacity zero:", leo_zero)
print("leo_capacity positive:", leo_pos)
print("apogee positive:", apogee_pos)
print("placeholders:", placeholders)
```

```python
print("=== leo_capacity == 0 ===")
for c in all_configs:
    if c.get("leo_capacity") == 0:
        print(f"{c['name']:35} apogee={c.get('apogee')}")

print("\n=== null leo_capacity, sample of 30 ===")
nulls = [c for c in all_configs if c.get("leo_capacity") is None]
for c in nulls[:30]:
    print(f"{c['name']:35} apogee={c.get('apogee')} launches={c.get('total_launch_count')}")
```
